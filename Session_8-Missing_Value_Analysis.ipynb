**How to create a Null values in a data**

- None
  
- np.nan

```python
d1 = {'Names':['Ramesh','Suresh',None,'Mahesh'],
     'Age':[20,22,24,None],
     'City':[None,'Blr','Pune','Mumbai']
    }

df1 = pd.DataFrame(d1)

df1
```

- Even though we provided None, the numerical column value by default **NaN**

- Categorical column value takes a **None**

```python
df1.dtypes
```

```python
df1.isnull()
```

```python
df1.isnull().sum()
```

```python
df1.to_csv('df1.csv',index=False)
```

```python
d2 = {'Names':['Ramesh','Suresh',np.nan,'Mahesh'],
     'Age':[20,22,24,np.nan],
     'City':[np.nan,'Blr','Pune','Mumbai']
    }

df2 = pd.DataFrame(d2)

df2
```

**None vs np.nan**

```python
arr1 = np.array([100,200,None])
arr2 = np.array([100,200,np.nan])
```

```python
sum(arr1)
```

```python
sum(arr2)
```

```python
100+None
```

```python
100+np.nan
```

```python
np.sum(arr2) # 100 + 200 + nan
```

```python
np.nansum(arr2) # 100 + 200
```
