Feature Scaling
------

- Feature Scaling is a method, to modify the value either increase or decrease

- In ML, we have different columns

    - age range between 0 to 100

    - height range between 0 to 200

    - income range between 10K to 1 Cr

- different columns having different range of values also different units

-  ML models might be think more values leads to more impact

-  Scale all the columns into a one scale is a better approach when model are working more on numbers

- Distance based algorithms are need to be scale

- Deep Learning main aim is lower the error : **Convergence**

- **Standard Scalar**

    - Also known as **Z Score**
      
    - 99.7 of data lies between -3 to 3
 
    - Mainly used in ML

- **MinMaxScaler**

    - Also known as **Normalization**
 
    - all the data lies between 0 to 1
 
    - Mainly used in image scaling
 
    - Generally color image values 0 to 255 RGB channel
 
    - Gray images ranges 0 to 1, 0 means Black and 1 means White

```python
# step 1: read the prevailing_wage data : wage_data (x)
# step 2: calculate mean
# step 3: calculate std
# step 4: Nr = s1 - s2
# step 5: Nr / s3
```

```python
wage_data = visa['prevailing_wage']
mean = wage_data.mean()
std = wage_data.std()
Nr = wage_data - mean
Z_wage = Nr / std
```

```python
Z_wage.hist()
```

```python
plt.subplot(2,2,1).hist(wage_data)
plt.subplot(2,2,2).hist(Z_wage)
plt.subplot(2,2,3).boxplot(wage_data)
plt.subplot(2,2,4).boxplot(Z_wage)
plt.show()
```
