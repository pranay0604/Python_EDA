**Emperical Rule**

```python
u-1*sigma to u+1*sigma ==== 68%

u-2*sigma to u+2*sigma ==== 95%

u-3*sigma to u+3*sigma ==== 99.7%


step-1: u

step-2: std

step-3: lb=u-1*std

step-4: ub= u+1*std

step-5: con1= wage_data>lb

step-6: con2=wage_data<ub

step-7: con = con1 & con2

step-8: len(visa_d[con])

step-9: 68%len (visa_df)
```

```python
u = wage_data.mean()
std_dev = wage_data.std()
lb = u - 1 * std_dev
ub = u + 1 * std_dev
con1 = wage_data > lb
con2 = wage_data < ub
con = con1 & con2
len(visa[con]), 68*len(visa)/100
```

```python
plt.figure(figsize=(12,3))
plt.subplot(1,2,1).hist(wage_data)
plt.subplot(1,2,2).boxplot(wage_data,vert=False)
plt.show()
```



          Q1-1.5IQR   Q1   median  Q3   Q3+1.5IQR
                       |-----:-----|
       o      |--------|     :     |--------|    o  o
                       |-----:-----|
     flier             <----------->            fliers

- Outliers are more than Q3 + 1.5 IQR

- Outliers are more than Q3 - 1.5 IQR

```python
# Step 1 : Q1
# Step 2 : Q2
# Step 3 : Q3
# Step 4 : IQR
# Step 5 : UB
# Step 6 : LB
# Step 7 : con1 
# Step 8 : con2 
# Step 9 : cond 
# Step 10 : outliers

Q1_25 = np.quantile(wage_data,0.25)
Q2_50 = np.quantile(wage_data,0.50)
Q3_75 = np.quantile(wage_data,0.75)
IQR = Q3_75 - Q1_25
UB = Q3_75 + (1.5*IQR)
LB = Q3_75 - (1.5*IQR)
con1 = wage_data < LB
con2 = wage_data > UB
cond = con1 | con2
outliers = wage_data[cond] 
len(outliers)
```

**WINSORIZATION**

- all more than UB values replace with UB
  
- all lower than LB values replace with LB

```python
wage_data = visa['prevailing_wage']
Q1_25 = np.quantile(wage_data,0.25)
Q2_50 = np.quantile(wage_data,0.50)
Q3_75 = np.quantile(wage_data,0.75)
IQR = Q3_75 - Q1_25
UB = Q3_75 + (1.5*IQR)
LB = Q3_75 - (1.5*IQR)

new_data_wage = wage_data.clip(LB,UB)
new_data_wage
```

```python
plt.figure(figsize=(12,3))
plt.subplot(1,2,1).hist(new_data_wage)
plt.subplot(1,2,2).boxplot(new_data_wage,vert=False)
plt.show()
```
