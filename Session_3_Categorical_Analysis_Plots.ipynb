```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

visa = pd.read_csv('/Users/pranay/Naresh IT/Data Files/Visadataset.csv')
```

```python
cat = visa.select_dtypes(include='object').columns
num = visa.select_dtypes(exclude='object').columns
```

```python
for i in cat[1:]:
    keys = visa[i].value_counts().keys()
    values = visa[i].value_counts().values
    d = pd.DataFrame(zip(keys,values),columns=['Lables','No of Applicants'])
    print(d)
    print()
    cwd = os.getcwd()
    folder_name = 'freq_table'
    os.makedirs(folder_name,exist_ok=True)
    new_dir = os.path.join(cwd,folder_name)
    path = os.path.join(new_dir,f"{i}.csv")
    
    d.to_csv(path,index=False)
    
```

**bar chart**

- x axis = Class

- y axis = No of Applicants

```python
# plt.bar(x,height,width)

keys = visa['continent'].value_counts().keys()
values = visa['continent'].value_counts().values
colors = ['Red','Blue','Green','Yellow','Orange','Violet']
plt.figure(figsize=(9,3))
plt.bar(keys,values,color=colors)
plt.xlabel("Class")
plt.ylabel("No of Applicants")
plt.title("Continent Bar Chart")
plt.show()
```
