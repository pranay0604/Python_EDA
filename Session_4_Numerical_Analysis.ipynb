```python
l = []
for i in num:
    data = visa[i]
    count = len(data)
    Min = np.min(data)
    p_25 = np.percentile(data,25)
    mean = np.mean(data)
    median = np.median(data)
    p_50 = np.percentile(data,50)
    p_75 = np.percentile(data,75)
    Max = np.max(data)
    std = np.std(data)

    data = [ count, Min, p_25, mean, median, p_50, p_75, Max, std]
    l.append(data)
    
idx = [ 'Count','Min','25P','Mean','Median','50P','75P','Max','Std']
df = pd.DataFrame(l,index=num,columns=idx)
df
```
