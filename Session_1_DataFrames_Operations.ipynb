**Step 13 : Concatenate two dataframes**

```python
df1 =  {'Names':'Suresh','Age':22,'City':'Pune'}
df2 =  {'Names':'Rames','Age':21,'City':'Hyd'}
pd.concat([df1,df2])
```

```python
names = ['Ramesh','Suresh','Satish']
age = [20,22,24]
data=zip(names,age)
df1 = pd.DataFrame(data)
name = ['Ram','Shyam','Laxman']
ages = [26,28,30]
data1=zip(name,ages)
df1 = pd.DataFrame(data1)
pd.concat([df1,df2])
```

```python
d1 =  {'Names':['Suresh'],'Age':[22],'City':['Pune']}
d2 =  {'Names':['Ramesh'],'Age':[21],'City':['Hyd']}
df1 = pd.DataFrame(d1)
df2 = pd.DataFrame(d2)
pd.concat([df1,df2],ignore_index=True)
```
