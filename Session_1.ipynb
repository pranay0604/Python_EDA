**Step 8 : Create a dataframe using dictionary**

- **keys** becomes columns
- **values** becomes rows

```python
data = {'Names':['Ramesh','Suresh','Satish'],
       'Age':[20,22,24],
       'City':['Hyd','Pune','Blr'],
       'Marks':[100,200,300]}
df = pd.DataFrame(data)
df
```
