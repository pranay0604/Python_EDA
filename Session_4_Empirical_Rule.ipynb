**Emperical Rule**

```python
u-1*sigma to u+1*sigma ==== 68%

u-2*sigma to u+2*sigma ==== 95%

u-3*sigma to u+3*sigma ==== 99.7%


step-1: u

step-2: std

step-3: lb=u-1*std

step-4: ub= u+1*std

step-5: con1= wage_data>lb

step-6: con2=wage_data<ub

step-7: con = con1 & con2

step-8: len(visa_d[con])

step-9: 68%len (visa_df)
```

```python
u = wage_data.mean()
std_dev = wage_data.std()
lb = u - 1 * std_dev
ub = u + 1 * std_dev
con1 = wage_data > lb
con2 = wage_data < ub
con = con1 & con2
len(visa[con]), 68*len(visa)/100
```

```python
for i in range(1,4):
    u = wage_data.mean()
    std_dev = wage_data.std()
    lb = u - i * std_dev
    ub = u + i * std_dev
    con1 = wage_data > lb
    con2 = wage_data < ub
    con = con1 & con2
    print(len(visa[con]))
```

```python
def Emperical(k,p):
    u = wage_data.mean()
    std_dev = wage_data.std()
    lb = u - k * std_dev
    ub = u + k * std_dev
    con1 = wage_data > lb
    con2 = wage_data < ub
    con = con1 & con2
    res=(len(visa[con]))==p*len(visa)/100
    return ({p:f"{res} : {[ len(visa[con]), p*len(visa)/100]}"})

# print(Emperical(1,68))
# print(Emperical(1,95))
# print(Emperical(1,99.7))

dict1 = Emperical(1,68)
dict1.update(Emperical(2,95))
dict1.update(Emperical(3,99.7))
dict1
```
