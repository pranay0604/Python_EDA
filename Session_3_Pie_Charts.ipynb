**Task**

```python
# Analyze sales data by regions and products, finding total sales, average sales, and the number of transactions per region. 
# Solve all below problems by using pandas.

# data = {
# "Region': [North', "South", 'East", "West", "North', "South,
# "East, West),
# "Product: 'A', 'A', 'B', 'B', 'A', 'B', 'A', 'B',
# 'Sales": [200, 300, 150, 400, 250, 100, 300, 500],
# "Transactions': [2, 4, 3, 5, 1, 2, 6, 3],
# }

# a. Total sales per region
# b. Count of transactions per region
# c. Average sales per region
# d. Total number of transactions per region
```

```python
data = {
'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'],
"Product": ['A', 'A', 'B', 'B', 'A', 'B', 'A', 'B'],
"Sales": [200, 300, 150, 400, 250, 100, 300, 500],
"Transactions": [2, 4, 3, 5, 1, 2, 6, 3]
}
```

```python
data_df = pd.DataFrame(data)
```

```python
data_df
```

```python
data_df.dtypes
```

```python
cat = data_df.select_dtypes(include='object').columns
num = data_df.select_dtypes(exclude='object').columns
```

```python
cat
```

```python
num
```

```python
for i in cat:
    keys = data_df[i].value_counts().keys()
    values = data_df[i].value_counts().values
    d = pd.DataFrame(zip(keys,values),columns=['Class','No'])
    print(d)
    plt.figure(figsize=(8,3))
    plt.bar(keys,values)
    plt.show()
```

```python
for i in num:
    keys = data_df[i].value_counts().keys()
    values = data_df[i].value_counts().values
    d = pd.DataFrame(zip(keys,values),columns=['Class','No'])
    print(d)
    plt.figure(figsize=(8,3))
    plt.bar(keys,values)
    plt.show()
```

```python
# a. Total sales per region

total_sales = data_df.groupby("Region")['Sales'].sum()
total_sales
```

```python
# b. Count of transactions per region

trans_region = data_df.groupby('Region')['Transactions'].size()
trans_region
```

```python
# c. Average sales per region

avg_sales = data_df.groupby('Region')['Sales'].mean()
avg_sales
```

```python
# d. Total number of transactions per region

total_trans = data_df.groupby('Region')['Transactions'].sum()
total_trans
```

```python
summary = data_df.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Sales_Records=("Region", "count"),
    Total_Transactions=("Transactions", "sum")
    )

print(summary)
```

```python
import os

new_folder = 'pie_charts'
os.makedirs(new_folder,exist_ok=True)
cwd = os.getcwd()
new_dir = os.path.join(cwd,new_folder)

for i in cat[1:]:
    keys = visa[i].value_counts().keys()
    values = visa[i].value_counts().values
    plt.pie(values,
            labels=keys,
            autopct='%1.2f%%',
            startangle=180,
            explode=[0.1 for i in range(len(values))],
            radius=1
           )
    path = os.path.join(new_dir, i)
    plt.savefig(f"{path}.png")
    plt.show()
```
